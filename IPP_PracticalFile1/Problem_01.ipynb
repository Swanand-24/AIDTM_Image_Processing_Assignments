## Problem Statement 1

Write a Python program to read an image named "flower.jpg" using OpenCV and display it in a window titled "Original Image".

```python
import cv2

image = cv2.imread("flower.jpg")
if image is None:
    raise FileNotFoundError("Could not find flower.jpg")
cv2.imshow("Original Image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
