## Problem Statement 2

Write a program to print the dimensions (height, width, channels) and data type of an image named "scene.jpg".

```python
import cv2

image = cv2.imread("scene.jpg")
if image is None:
    raise FileNotFoundError("Could not find scene.jpg")
height, width, channels = image.shape
print("Height:", height)
print("Width:", width)
print("Channels:", channels)
print("Data type:", image.dtype)
```
