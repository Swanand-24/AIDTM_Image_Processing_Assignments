## Problem Statement 3

Write a Python program to load a colour image and convert it into a grayscale image, then save the result as "output_gray.jpg".

```python
import cv2

image = cv2.imread("flower.jpg")
if image is None:
    raise FileNotFoundError("Could not find flower.jpg")
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2.imwrite("output_gray.jpg", gray_image)
print("Grayscale image saved as output_gray.jpg")
```
