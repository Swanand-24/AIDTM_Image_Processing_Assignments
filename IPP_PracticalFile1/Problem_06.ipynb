## Problem Statement 6

Write a Python code snippet to crop the central 200×200 pixel area of an image and display it using Matplotlib.

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("flower.jpg")
if image is None:
    raise FileNotFoundError("Could not find flower.jpg")
height, width = image.shape[:2]
if height < 200 or width < 200:
    raise ValueError("Image must be at least 200×200 pixels")
x = (width - 200) // 2
y = (height - 200) // 2
cropped_image = image[y:y + 200, x:x + 200]
cropped_image = cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB)
plt.imshow(cropped_image)
plt.axis("off")
plt.show()
```
