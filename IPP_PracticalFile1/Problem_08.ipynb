## Problem Statement 8

Write a Python statement to flip the image "car.jpg" left-to-right using OpenCV and display the result.

```python
import cv2

image = cv2.imread("car.jpg")
if image is None:
    raise FileNotFoundError("Could not find car.jpg")
flipped_image = cv2.flip(image, 1)
cv2.imshow("Flipped Image", flipped_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
