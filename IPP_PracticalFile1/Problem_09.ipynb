## Problem Statement 9

Write a Python program that converts an image from RGB to HSV colour space, and display each channel (Hue, Saturation, Value) separately using Matplotlib.

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("flower.jpg")
if image is None:
    raise FileNotFoundError("Could not find flower.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
hsv_image = cv2.cvtColor(image, cv2.COLOR_RGB2HSV)
hue, saturation, value = cv2.split(hsv_image)
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(hue, cmap="gray")
axes[0].set_title("Hue")
axes[1].imshow(saturation, cmap="gray")
axes[1].set_title("Saturation")
axes[2].imshow(value, cmap="gray")
axes[2].set_title("Value")
for axis in axes:
    axis.axis("off")
plt.tight_layout()
plt.show()
```
