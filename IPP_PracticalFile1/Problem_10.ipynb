## Problem Statement 10

Write a short Python code snippet to compute the Mean Squared Error (MSE) between two grayscale images, "img1.jpg" and "img2.jpg".

```python
import cv2
import numpy as np

image1 = cv2.imread("img1.jpg", cv2.IMREAD_GRAYSCALE)
image2 = cv2.imread("img2.jpg", cv2.IMREAD_GRAYSCALE)

if image1 is None:
    raise FileNotFoundError("Could not find img1.jpg")
if image2 is None:
    raise FileNotFoundError("Could not find img2.jpg")

image2 = cv2.resize(image2, (image1.shape[1], image1.shape[0]))

mse = np.mean((image1.astype(np.float64) - image2.astype(np.float64)) ** 2)

print("Image 1 Dimensions:", image1.shape)
print("Image 2 Dimensions:", image2.shape)
print("MSE:", mse)
```
